```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# 1. Carregando o BERTimbau otimizado para geração de embeddings de sentenças
print("Carregando o modelo BERTimbau...")
model = SentenceTransformer('neuralmind/bert-base-portuguese-cased')

# 2. Definindo a base de conhecimento do Partido (Regulamentos e Diretrizes Oficiais)
# Na sua aplicação real, aqui entram os trechos dos estatutos e notas oficiais do partido.
documentos_partido = [
    "Defendemos a rigorosa austeridade fiscal, controle de gastos públicos e privatizações de estatais.",
    "O Estado deve intervir fortemente na economia para garantir programas sociais de transferência de renda.",
    "Apoio irrestrito à desregulamentação de mercados e redução de impostos para o setor produtivo."
]

# 3. Definindo os discursos dos Parlamentares a serem testados
discursos_parlamentares = {
    "Parlamentar A (Opositor/Divergente)": "Precisamos ampliar urgentemente os investimentos estatais e proteger as empresas públicas contra qualquer venda.",
    "Parlamentar B (Aliado/Alinhado)": "O caminho para o crescimento é o corte de gastos do governo e a abertura para a iniciativa privada atuar."
}

# 4. Gerando os Vetores (Embeddings)
print("Gerando vetores...")
vetor_partido = model.encode(documentos_partido)
# Tiramos a média dos vetores dos documentos do partido para criar um "centróide ideológico" do partido
vetor_centroid_partido = np.mean(vetor_partido, axis=0).reshape(1, -1)

# 5. Calculando a Similaridade (Alinhamento Político)
print("\n--- RESULTADOS DO ALINHAMENTO PARTIDÁRIO ---")
for nome, discurso in discursos_parlamentares.items():
    vetor_discurso = model.encode([discurso])

    # Similaridade de Cosseno (varia de -1 ou 0 até 1. Quanto mais próximo de 1, mais alinhado)
    similaridade = cosine_similarity(vetor_discurso, vetor_centroid_partido)[0][0]

    print(f"\nOrador: {nome}")
    print(f"Discurso: \"{discurso}\"")
    print(f"Grau de Alinhamento com o Partido (0 a 1): {similaridade:.4f}")
```
